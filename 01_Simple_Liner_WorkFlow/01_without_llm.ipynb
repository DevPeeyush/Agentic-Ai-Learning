## BMI Calculator using Non-LLM workflow in lang grapgh

`Input`--> `call BMI ` --> `Output`
- ` State ` ---> weight, height, bmi_value

```python
from langgraph.graph import StateGraph, START,END
from typing import TypedDict
```

```python
# Define State
class BMIState(TypedDict):
    weight_kg: float
    height_m: float
    bmi: float
    category: str
```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    # define calculate_bmi(state: BMIState (it is input)) -> BMIState:(this gives us output)
    weight = state['weight_kg']
    height = state['height_m']
    
    bmi = weight / (height ** 2)
    
    state['bmi'] = round(bmi, 2)
    return state  
```

```python
def label_bmi(state: BMIState) -> BMIState:
    bmi = state['bmi']
    
    if bmi < 18.5:
        state['category'] = 'Underweight'
    elif 18.5 <= bmi < 24.9:
        state['category'] = 'Normal weight'
    elif 25 <= bmi < 29.9:
        state['category'] = 'Overweight'
    else:
        state['category'] = 'Obesity'
    
    return state
    
```

```python
# Define our graph
graph = StateGraph(BMIState)

# Adding nodes to our graph
graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('Label_BMI',label_bmi)

# Adding edges to our graph
graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', 'Label_BMI')
graph.add_edge('Label_BMI', END)
# compile the graph
workflow = graph.compile()
```

```python
# execute the graph
result = workflow.invoke({'weight_kg': 90, 'height_m': 1.75})
print(result)  # Output
```
